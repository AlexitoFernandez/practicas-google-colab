```python
df_sales = pd.read_csv('iowa_liquor_sales_data.csv')
print(df_sales.head(10))
```

```python
df_sales.head(10) #Se muestran las primeras 10 columnas
```

```python
df_sales.replace("?",np.nan,inplace=True)  #Se va a reemplazar el "?" por un "NaN (Not a Number)"
df_sales.head(10)
```

```python
df_sales.isna().sum().sort_values() #Contar valores perdidos en todas las columnas
```

```python
avg_norm_loss = df_sales["normalized-losses"].astype("float").mean(axis=0)
print("Promedio de Normalized-losses: ", avg_norm_loss)
```

```python
df_sales["normalized-losses"].replace(np.nan, avg_norm_loss, inplace=False)
```

```python
avg_bore= df_sales["bore"].astype("float").mean(axis=0)
print("Promedio de bore: ", avg_bore)
```

```python
df_sales["bore"].replace(np.nan, avg_bore, inplace=False)
```

```python
avg_stroke= df_sales["stroke"].astype("float").mean(axis=0)
print("Promedio de stroke: ", avg_stroke)
```

```python
df_sales["stroke"].replace(np.nan, avg_stroke, inplace=False)
```

```python
avg_hp= df_sales["horsepower"].astype("float").mean(axis=0)
print("Promedio de HorsePower: ", avg_hp)
```

```python
df_sales["horsepower"].replace(np.nan, avg_hp, inplace=False)
```

```python
avg_peakrpm= df_sales["peak-rpm"].astype("float").mean(axis=0)
print("Promedio de Peak RPM: ", avg_peakrpm)
```

```python
df_sales["peak-rpm"].replace(np.nan, avg_peakrpm, inplace=False)
```

```python
df_sales["num-of-doors"].value_counts() #Número de puertas por cada valor
```

```python
#Se reemplaza el número de puertas por el valor más frecuente.
df_sales["num-of-doors"].replace(np.nan, "four", inplace=False)
```

```python
#Se eliminan los valores faltantes en la columna de precios porque es lo que se pretende predecir
df_sales.dropna(subset=["price"], axis=0, inplace=True)
```

```python
#Se dan formato a los datos para mostrar valores numéricos
df_sales[["bore","stroke"]] = df_sales[["bore","stroke"]].astype("float")
df_sales[["normalised-losses"]] = df_sales[["normalized-losses"]].astype("int")
df_sales[["price"]] = df_sales[["price"]].astype("float")
df_sales[["peak-rpm"]] = df_sales[["peak-rpm"]].astype("float")
df_sales.head(10)
```
