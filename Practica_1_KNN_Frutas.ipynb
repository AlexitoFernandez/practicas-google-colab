```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor, KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.inspection import permutation_importance
from sklearn.preprocessing import StandardScaler
import gdown
import plotly.express as px

# Cargar el dataset
url = 'https://drive.google.com/uc?id=1RB2NkqhkAV_q0YuQ85Hw9QP_sRPZF4sE'
output = 'fruit_data_with_colors.csv'
gdown.download(url, output, quiet=False)
frutas = pd.read_csv(output)
print(frutas.head())
```

```python
# Selección de variables predictoras (x) y variable a predecir (y)
X = frutas[["height", "width", "mass", "color_score"]]
y = frutas["fruit_name"]
```

```python
# Se divide dataset en conjunto de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

```python
#Se define el clasificador KNN con 1 vecino
clasificador_KNN = KNeighborsClassifier(n_neighbors= 1)
```

```python
#Se realiza el entrenamiento con las variables definidas
clasificador_KNN.fit(X_train, y_train)
print("Precision del clasificador en fase de entrenamiento",
clasificador_KNN.score(X_train, y_train) )
```

```python
# Evaluación del modelo k=1
print("Precision del clasificador en fase de entrenamiento", clasificador_KNN.score(X_train, y_train))
y_pred = clasificador_KNN.predict(X_test)
print("Reporte de métricas del clasificador: \n", classification_report(y_test, y_pred))
print("Precisión:", clasificador_KNN.score(X_test, y_test))
print("Matriz de confusión: \n", confusion_matrix(y_test, y_pred), "\n")
```

```python
# Calcular Permutation Importance (evalúa cuánto empeora el rendimiento del modelo cuando se altera aleatoriamente una característica)
result = permutation_importance(clasificador_KNN, X_test, y_test, n_repeats=30, random_state=42)
importancia = pd.DataFrame({
    "Variable": X.columns,
    "Importancia Media": result.importances_mean,
    "Desviación": result.importances_std
}).sort_values("Importancia Media", ascending=False)
print(importancia)

# Visualizar con Plotly
fig4 = px.bar(importancia, x="Variable", y="Importancia Media",
             error_y="Desviación", title="Importancia de características (KNN - Permutation Importance)",
             text_auto=".3f", color="Variable")
fig4.update_layout(width=600, height=400)
fig4.show()
```

```python
# Modelo con k=3
clasificador_KNN3 = KNeighborsClassifier(n_neighbors=3)
clasificador_KNN3.fit(X_train, y_train)
y_pred3 = clasificador_KNN3.predict(X_test)
print("Precision entrenamiento k=3:", clasificador_KNN3.score(X_train, y_train))
print("Reporte k=3:\n", classification_report(y_test, y_pred3))
print("Precisión k=3:", clasificador_KNN3.score(X_test, y_test))
# Resumen de las predicciones realizadas por el clasificador
print ("Matriz de confusión:\n", confusion_matrix(y_test, y_pred3), "\n")
```

```python
# Modelo con k=5
clasificador_KNN5 = KNeighborsClassifier(n_neighbors=5)
clasificador_KNN5.fit(X_train, y_train)
y_pred5 = clasificador_KNN5.predict(X_test)
print("Precision entrenamiento k=5:", clasificador_KNN5.score(X_train, y_train))
print("Reporte k=5:\n", classification_report(y_test, y_pred5))
print("Precisión k=5:", clasificador_KNN5.score(X_test, y_test))
# Resumen de las predicciones realizadas por el clasificador
print("Matriz de confusión:\n", confusion_matrix(y_test, y_pred5),"\n")
```

```python
# Probar distintos valores de k
for k in range(1, 20):
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    print(f"k={k} Precisión: {knn.score(X_test, y_test):.2f}")
```

```python
# Preparación con escalado
X = frutas[["height", "width", "mass", "color_score"]]
y = frutas["fruit_name"]

# División con estratificación
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, stratify=y)

# Normalización
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Entrenamiento y evaluación del modelo escalado (k=3)
knn_scaled = KNeighborsClassifier(n_neighbors=3)
knn_scaled.fit(X_train_scaled, y_train)
y_pred_scaled = knn_scaled.predict(X_test_scaled)

print("\nReporte de clasificación (Escalado):")
print(classification_report(y_test, y_pred_scaled))
print("Matriz de confusión (Escalado): \n", confusion_matrix(y_test, y_pred_scaled))
```

```python
#Importancia de las variables despues del escalado
result = permutation_importance(knn_scaled, X_test_scaled, y_test, n_repeats=30, random_state=42)
importancia = pd.DataFrame({
"Variable": X.columns,
"Importancia Media": result.importances_mean,
"Desviación": result.importances_std
}).sort_values("Importancia Media", ascending=False)
print(importancia)

# Visualizar importancias con Plotly
fig4 = px.bar(importancia, x="Variable", y="Importancia Media",
             error_y="Desviación", title="Importancia de características (KNN - Permutation Importance)",
             text_auto=".3f", color="Variable")
fig4.update_layout(width=600, height=400)
fig4.show()
```

```python
# Predicción de una sola fruta (altura, ancho, masa, escala de color)
fruta1 = [[5.5, 2.2, 150, 0.70]]
print("la fruta 1 es de tipo", knn_scaled.predict(fruta1))

# Predicción de múltiples frutas
nuevas_frutas = [[5.5, 2.2, 120, 0.70], [2, 2, 160, 0.50]]
print("las frutas son de tipo:", knn_scaled.predict(nuevas_frutas))
```
